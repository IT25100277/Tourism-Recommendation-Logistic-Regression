```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('tourism_recommendation_dataset_en.csv')

# Scale numerical features to a range between 0 and 1
scaler = MinMaxScaler()
scaled_cols = ['ticket_price_scaled', 'spend_amount_scaled', 'duration_scaled']
df[scaled_cols] = scaler.fit_transform(df[['ticket_price', 'spend_amount', 'visit_duration_hours']])

# Visualize the data distributions
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df['ticket_price_scaled'], ax=axes[0], color='teal', kde=True)
axes[0].set_title('Ticket Price Scaled [0, 1]')
sns.histplot(df['spend_amount_scaled'], ax=axes[1], color='purple', kde=True)
axes[1].set_title('Spend Amount Scaled [0, 1]')
sns.histplot(df['duration_scaled'], ax=axes[2], color='orange', kde=True)
axes[2].set_title('Visit Duration Scaled [0, 1]')
plt.savefig('IT25100277_scaled_histograms.png', bbox_inches='tight')
plt.show()
```
